```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Create a blank image
image = np.zeros((500, 500, 3), dtype=np.uint8)
```

```python
# Add text on the image using cv2.putText
font = cv2.FONT_HERSHEY_SIMPLEX
cv2.putText(image, 'Open and Close', (100, 250), font, 1, (255, 255, 255), 2, cv2.LINE_AA)
```

```python
# Create a simple square kernel (3x3)
kernel = np.ones((3, 3), np.uint8)
```

```python
# Display the input image
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))  # Convert BGR to RGB for displaying
plt.title("Input Image with Text")
plt.axis('off')
```

```python
# Opening is erosion followed by dilation
opened_image = cv2.morphologyEx(image, cv2.MORPH_OPEN, kernel)
```

```python
# Display the result of Opening
plt.imshow(cv2.cvtColor(opened_image, cv2.COLOR_BGR2RGB))  # Convert BGR to RGB
plt.title("Opening Operation")
plt.axis('off')
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Create black image
img = np.zeros((300, 500), dtype=np.uint8)

# Put text on image
cv2.putText(img, "Open and Close",
            (80, 150),
            cv2.FONT_HERSHEY_SIMPLEX,
            1.2,
            (255),
            2,
            cv2.LINE_AA)

# Add noise (so morphology has effect)
noise = np.random.randint(0, 2, (300, 500)) * 255
noisy_img = cv2.bitwise_or(img, noise.astype(np.uint8))

# Kernel
kernel = np.ones((5,5), np.uint8)

# Opening (remove noise)
opened = cv2.morphologyEx(noisy_img, cv2.MORPH_OPEN, kernel)

# Closing (fill gaps)
closed = cv2.morphologyEx(noisy_img, cv2.MORPH_CLOSE, kernel)
```

```python
plt.figure(figsize=(10,6))

plt.subplot(1,3,1)
plt.imshow(noisy_img, cmap='gray')
plt.title("Noisy Image")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(opened, cmap='gray')
plt.title("Opening Operation")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(closed, cmap='gray')
plt.title("Closing Operation")
plt.axis('off')

plt.show()
```
